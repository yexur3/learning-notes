```python
import sionna.rt as rt
import numpy as np
from importlib.metadata import version

print("Sionna RT:", version("sionna-rt"))
```

```python
terrain_scene = rt.load_scene(
    rt.scene.san_francisco,
    merge_shapes=False
)

print("Count of objects:", len(terrain_scene.objects))
print("First 15 names:", list(terrain_scene.objects)[:15])
```

```python
terrain_scene.frequency = 3.5e9

terrain_scene.tx_array = rt.PlanarArray(
    num_cols=1,
    num_rows=1,
    pattern="iso",
    polarization = "V"
)

terrain_scene.rx_array = terrain_scene.tx_array

picked_point = np.array([612.388, 76.674, 151.618])

terrain_tx = rt.Transmitter(
    name = "hill_tx",
    position=(picked_point + [0,0,3]).tolist(),
    power_dbm=30,
    display_radius=3
)

terrain_rx = rt.Receiver(
    name = "square_rx",
    position=[228.919, -108.567, 23.5],
    display_radius=3
)

terrain_scene.add(terrain_rx)
terrain_scene.add(terrain_tx)

terrain_solver = rt.PathSolver()

los_paths = terrain_solver(
    scene=terrain_scene,
    max_depth=0,
    los=True,
    refraction=False
)

reflection_paths = terrain_solver(
    scene=terrain_scene,
    max_depth = 1,
    los=True,
    specular_reflection=True,
    diffuse_reflection=False,
    refraction=False
)

print("Valid paths:", los_paths.valid)
print("Valid paths:", reflection_paths.valid)

terrain_scene.preview(paths=reflection_paths)
```
