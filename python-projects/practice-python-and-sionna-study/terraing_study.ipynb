```python
import sionna.rt as rt
import numpy as np
from importlib.metadata import version

print("Sionna RT:", version("sionna-rt"))
```

```python
terrain_scene = rt.load_scene(
    rt.scene.san_francisco,
    merge_shapes=False
)

print("Count of objects:", len(terrain_scene.objects))
print("First 15 names:", list(terrain_scene.objects)[:15])
```

```python
terrain_scene.frequency = 3.5e9

terrain_scene.tx_array = rt.PlanarArray(
    num_cols=1,
    num_rows=1,
    pattern="iso",
    polarization = "V"
)

terrain_scene.rx_array = terrain_scene.tx_array

picked_point = np.array([612.388, 76.674, 151.618])

terrain_tx = rt.Transmitter(
    name = "hill_tx",
    position=(picked_point + [0,0,3]).tolist(),
    power_dbm=30,
    display_radius=3
)

terrain_rx = rt.Receiver(
    name = "square_rx",
    position=[228.919, -108.567, 23.5],
    display_radius=3
)

terrain_scene.add(terrain_rx)
terrain_scene.add(terrain_tx)

terrain_solver = rt.PathSolver()

los_paths = terrain_solver(
    scene=terrain_scene,
    max_depth=0,
    los=True,
    refraction=False
)

reflection_paths = terrain_solver(
    scene=terrain_scene,
    max_depth = 1,
    los=True,
    specular_reflection=True,
    diffuse_reflection=False,
    refraction=False
)

print("Valid paths:", los_paths.valid)
print("Valid paths:", reflection_paths.valid)

terrain_scene.preview(paths=reflection_paths)
```

```python
terrain_names = [
    name for name in terrain_scene.objects
    if "terrain" in name.lower()
]

print(terrain_names)

ground = terrain_scene.get("Terrain")

print(ground.radio_material)
```

```python
original_ground_material = ground.radio_material
print(original_ground_material)
```

```python
metal_ground = rt.ITURadioMaterial(
    name = "experimental_metal_ground",
    itu_type = "metal",
    thickness = 0.1
)

ground.radio_material = metal_ground

metal_paths = terrain_solver(
    scene=terrain_scene,
    max_depth = 1,
    los=True,
    specular_reflection=True,
    diffuse_reflection=False,
    refraction=False
)

terrain_scene.preview(paths=metal_paths)
```

```python
a_before, _ = reflection_paths.cir(
    normalize_delays=False,
    out_type="numpy"
)

a_after, _ = metal_paths.cir(
    normalize_delays=False,
    out_type = "numpy"
)

print("Concrete:", np.abs(a_before).reshape(-1) ** 2)
print("Metal:", np.abs(a_after).reshape(-1) ** 2)
```
